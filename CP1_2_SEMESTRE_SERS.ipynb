##8.Explique por que esse segundo conjunto de registros pode merecer maior atenção da equipe de energia.

O segundo conjunto de registros, representado pelo `novo_dataframe`, que contém simultaneamente um **consumo de energia (Consumo_kWh) elevado** (acima de 75% do máximo) e um **fator de potência (Fator_potencia_3) baixo** (abaixo de 5), merece uma atenção especial da equipe de energia por várias razões:

1.  **Ineficiência Energética**: Um baixo fator de potência indica que o sistema elétrico está utilizando a energia de forma ineficiente. Uma parte significativa da energia fornecida é reativa (não realiza trabalho útil), em vez de ativa (que de fato realiza trabalho). Combinar isso com um consumo elevado significa que há um grande desperdício potencial de energia.

2.  **Custos Operacionais Elevados**: Companhias elétricas frequentemente impõem multas ou taxas adicionais para consumidores com baixo fator de potência devido ao impacto negativo na rede. Portanto, esses registros podem indicar um custo de eletricidade desnecessariamente alto.

3.  **Sobrecarga e Perdas na Rede**: Um baixo fator de potência aumenta a corrente total na rede para a mesma quantidade de energia ativa. Isso leva a:
    *   **Maiores perdas (I²R)** nos cabos e transformadores, resultando em calor e desperdício de energia.
    *   **Redução da capacidade** dos equipamentos elétricos, pois eles precisam lidar com correntes mais altas.
    *   **Quedas de tensão**, que podem afetar o desempenho de outros equipamentos.

4.  **Risco de Problemas Operacionais**: A ineficiência e as maiores correntes podem levar ao superaquecimento de equipamentos, reduzindo sua vida útil e aumentando o risco de falhas ou paradas não programadas.

Ao identificar esses períodos de alto consumo e baixo fator de potência, a equipe de energia pode investigar as causas subjacentes (como equipamentos descalibrados, motores ineficientes, ou falta de capacitores para correção do fator de potência) e implementar medidas corretivas para otimizar o uso da energia, reduzir custos e melhorar a estabilidade da rede.

 # 7.	Explique por que não seria adequado utilizar diretamente o mesmo valor numérico de potência como limite para as duas fontes sem antes observar a escala de cada variável.
 Não seria adequado usar o mesmo valor numérico absoluto de potência como limite para as produções solar e eólica sem antes observar a escala de cada uma, e aqui está o porquê:

Diferença Inerente de Escala: Como vimos nos dados, a produção máxima de energia solar é de 16316 e a produção máxima de energia eólica é de 22929. Estas são escalas diferentes. Um valor limite de, digamos, 15000 unidades de energia, representaria uma proporção muito diferente do máximo para cada fonte.

Viés e Comparação Injusta: Se usássemos um limite fixo, estaríamos impondo uma régua que pode ser desproporcional. Por exemplo:

Um limite de 15000 para a energia solar seria um patamar muito alto (quase 92% do seu máximo), resultando em poucos registros de "alta geração".
Para a energia eólica, 15000 seria um patamar mais baixo (cerca de 65% do seu máximo), o que significaria mais registros de "alta geração" apenas porque seu potencial máximo é maior.

### 7. Comparação de DataFrames e o Efeito da Inclusão da Corrente como Segunda Condição
